2️⃣ Memory Model

````python
a = 5
b = a
a = 10
print(id(a))
print(id(b))
````

What happened?

5 object created

a → 5

b → 5

a now points to new object 10

b still points to 5


Key Rule: Python variables do not store values — they store references.

4️⃣ Why This Matters in ML / DL

Bad understanding = silent bugs 😈

````python
def update(weights):
    weights.append(0)

w = [1, 2, 3]
update(w)
print(w)
````

➡️ Model bug if you don’t expect mutation.

5️⃣ Industry-Level Rule

Never modify inputs unless explicitly required

````python
def update(weights):
    return weights + [0]

w = [1, 2, 3]
update(w)
print(w)
````

````python
a = 1000
b = 1000
print(a is b)
````

Explanation:

is checks object identity, not value

Small integers (-5 to 256) are cached

1000 is not guaranteed to be cached

````python
def scale(data, factor):
    for i in range(len(data)):
        data[i] *= factor
````

Why it’s dangerous

* It modifies the original data

**If the same dataset is reused:**

* Training data gets corrupted

* Validation leaks information

* Bugs become silent and hard to trace

* In ML, silent mutation = wrong models.

````python
def scale(data, factor):
    result = []
    for value in data:
        result.append(value * factor)
    return result
````

🧠 Reference vs Value Semantics

Value Semantics (NOT Python default)

Variable stores actual value

Copy happens automatically

Changes don’t affect others

````python
x = 5
y = x
x = 10
# y is still 5
````

````python
x = [1, 2]
y = x
y.append(3)
print(x)  # [1, 2, 3]
````

````python
# Key Rule

# Assignment never copies objects in Python.

# Only references are copied.
````
